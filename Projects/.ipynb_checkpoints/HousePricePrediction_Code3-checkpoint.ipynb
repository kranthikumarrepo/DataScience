## FeatureEncodingScaling
[GoBackToTop](#ProjectRequirement)

#### Feature Encoding
- Since ML algorithms are designed to work with numerical data, we need to convert non-numerical data to numeric, by performing encoding.
- If hierarchical categorical data, perform Ordinal Encoding and if nominal categorical data, perform OneHotEncoding.

#### Feature Scaling
- Perform Feature scaling to ensure that all features contribute equally to the model training process, preventing features with larger values from dominating the results.
- Since we might use XGB - gradient-based optimization, can be heavily influenced by features with larger values. Feature scaling ensures that no single feature inappropriately biases the model's learning process.
- Perform Standardization Scaling as we are going to deal with Linear Regression, Random Forest and Gradient Boosting.

#### Steps
- First identify all the Numerical and Categorical features separately.
- For Categorical features, identify all the Ordinal and Nominal Features separately.
- Once identified, perform Scaling and Encoding using Column Transformers and Pipelines.

```python
# Get all the numeric Features - as we have to perform scaling
# Store the list in a variable, this can be used in Column Transformer later on.

Numerical_Features = TrainDF_HousingData.select_dtypes(include=['int64','float64']).columns
Numerical_Features
```

```python
# Remove the output Feature - SalePrice from the Scaling list

Numerical_Features = Numerical_Features.drop(['SalePrice'])
Numerical_Features
```

```python
# Get the names of all the Categorical Features - We have to perform either One Hot Encoding or Ordinal Encoding depending on the type of data.
# Features that hold hierarchical data >>> Perform Ordinal encoding
# Features that do not have hierarchical data >>> Perform OneHotEncoding

TrainDF_HousingData.select_dtypes(include='object').columns
# Identify nominal and ordinal features based on the description of the features.
```

```python
# Make a list of only those features that require Ordinal Encoding.
# Store the list in a variable, this can be used in Column Transformer later on.

Ordinal_Encoding_Features = ['LotShape', 'LandContour','Utilities','LandSlope',  'BsmtQual',  'BsmtFinType1', 'BsmtFinType2', 'CentralAir',  \
                             'Functional', 'FireplaceQu', 'GarageFinish', 'GarageQual', 'PavedDrive', 'ExterCond', 'KitchenQual', 'BsmtExposure', \
                             'HeatingQC','ExterQual', 'BsmtCond','GarageCond']
```

```python
# Make a list of only those features that require OneHot Encoding.
# Store the list in a variable, this can be used in Column Transformer later on.

OneHot_Encoding_Features = ['Street', 'LotConfig','Neighborhood', 'Condition1', 'Condition2', 'BldgType', 'HouseStyle', 'RoofStyle', 'Exterior1st', \
                  'Exterior2nd', 'MasVnrType','Foundation',  'Electrical',  'SaleType', 'MSZoning', 'SaleCondition', 'Heating', 'GarageType', \
                  'RoofMatl']
```

#### Create Pipeline for numerical and categorical features
- The pipelines are used to handle multiple steps related to numerical and categorical features.

```python
# Create a pipeline for numerical features - to perform imputing with mean and then to perform Standard Scaling

Numerical_Features_PipelineObj = Pipeline(steps = [
    ('Imputer_NumericaFeatures_Mean',SimpleImputer(strategy='mean')), # Replace the nulls with Mean
    ('StandardScaler_NumericFeatures',StandardScaler()) # X_scaled = (X - mean) / standard_deviation.
])

# Standard Scaler - Scales the data to have a mean of 0 and a standard deviation of 1. i.e. once the standardization is done, the mean of 
# all the converted values will be 0 and SD = 1.
```

```python
OneHotEncoder
```

```python
# Create a pipeline for Ordinal features - to perform imputing with mode and then to perform Ordinal Encoding

Ordinal_Encoding_Features_PipelineObj = Pipeline(steps = [
    ('Imputer_Categorical_Ordinal_Mode',SimpleImputer(strategy='most_frequent')), # Replace the nulls with Mode
    ('OrdinalEncoder_Categorical',OrdinalEncoder(handle_unknown='use_encoded_value',unknown_value=-1))
])

# OrdinalEncoder_Obj = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']],dtype=np.int32)
# Generally we mention the categories and their priority, if we do not mention the categories parameter, it considers - 'auto' : Determine 
# categories automatically from the training data.
# handle_unknown = 'use_encoded_value' >>> in case an unknown categorical feature is present during transform, the encoded value of unknown 
# categories will be set to the value given for the parameter `unknown_value`.
```

```python
# Create a pipeline for Ordinal features - to perform imputing with mode and then to perform OneHot Encoding

OneHot_Encoding_Features_PipelineObj = Pipeline(steps = [
    ('Imputer_Categorical_Nominal_Mode',SimpleImputer(strategy='most_frequent')), # Replace the nulls with Mode
    ('OneHotEncoder_Categorical',OneHotEncoder(handle_unknown='ignore',sparse_output=False))
])

# handle_unknown='ignore' - When an unknown category is encountered during transform, the resulting one-hot encoded columns for this feature will 
# be all zeros. In the inverse transform, an unknown category will be denoted as None.
```

#### Column Transformers
- Mention the Column(Feature) Names and the type of action to be performed on each Feature

```python
# Now provide the list of Columns and the respective action to be performed on each of those columns.

ColumnTransformer_Obj = ColumnTransformer( transformers=[
    ('NumTransfPipeline',Numerical_Features_PipelineObj,Numerical_Features),
    ('OrdTransfPipeline',Ordinal_Encoding_Features_PipelineObj,Ordinal_Encoding_Features),
    ('OHETransfPipeline',OneHot_Encoding_Features_PipelineObj,OneHot_Encoding_Features)
], remainder='passthrough',n_jobs=-1)

# By default, only the specified columns in transformers are transformed and combined in the output, and the non-specified columns are dropped. 
# hence use remainder='passthrough' to pass any unknown columns, instead of dropping them
# n_jobs = -1 >>> Number of jobs to run in parallel >>> -1 = Use all the available processors 
```

#### Final Pipeline

```python
# This piepline will call the ColumnTransformer_Obj, which inturn calls the other pipelines.

Pipeline_AllCombined = Pipeline(steps = [
    ('FinalPreprocessingPipeline',ColumnTransformer_Obj)
])
```
